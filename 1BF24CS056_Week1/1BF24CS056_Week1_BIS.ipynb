```python
import random
import math

points = {
    0: (2, 3),
    1: (5, 8),
    2: (9, 2),
    3: (12, 7),
    4: (15, 3),
    5: (18, 9),
    6: (21, 4),
    7: (25, 8),
    8: (28, 3),
    9: (31, 7)
}

POPULATION_SIZE = 300
GENERATIONS = 200
MUTATION_RATE = 0.1
CROSSOVER_RATE = 0.5


def distance(a, b):
    x1, y1 = points[a]
    x2, y2 = points[b]

    return math.sqrt((x2 - x1) ** 2 + (y2 - y1) ** 2)


def route_distance(route):
    total = 0

    for i in range(len(route)):
        total += distance(route[i], route[(i + 1) % len(route)])

    return total


def create_population():
    population = []
    cities = list(points.keys())

    for _ in range(POPULATION_SIZE):
        route = cities.copy()
        random.shuffle(route)
        population.append(route)

    return population


def tournament_selection(population):
    selected = []

    for _ in range(POPULATION_SIZE // 2):
        a = random.choice(population)
        b = random.choice(population)

        if route_distance(a) < route_distance(b):
            selected.append(a)
        else:
            selected.append(b)

    return selected


def rank_selection(population):
    ranked = sorted(population, key=route_distance)

    weights = list(range(len(ranked), 0, -1))

    selected = random.choices(
        ranked,
        weights=weights,
        k=POPULATION_SIZE // 2
    )

    return selected


def pmx(parent1, parent2):
    n = len(parent1)

    start, end = sorted(random.sample(range(n), 2))

    child = [None] * n

    child[start:end] = parent1[start:end]

    for i in range(start, end):
        if parent2[i] not in child:

            position = i

            while start <= position < end:
                value = parent1[position]
                position = parent2.index(value)

            child[position] = parent2[i]

    for i in range(n):
        if child[i] is None:
            child[i] = parent2[i]

    return child


def eos(route):
    child = route.copy()

    i, j = random.sample(range(len(child)), 2)

    child[i], child[j] = child[j], child[i]

    return child


def eom(route):
    child = route.copy()

    i, j = sorted(random.sample(range(len(child)), 2))

    child[i:j + 1] = reversed(child[i:j + 1])

    return child


def create_new_population(parents, large_problem):

    new_population = parents.copy()

    while len(new_population) < POPULATION_SIZE:

        parent1 = random.choice(parents)
        parent2 = random.choice(parents)

        if random.random() < CROSSOVER_RATE:
            child = pmx(parent1, parent2)
        else:
            child = parent1.copy()

        if random.random() < MUTATION_RATE:

            if large_problem:
                child = eom(child)
            else:
                child = eos(child)

        new_population.append(child)

    return new_population


population = create_population()

N = len(points)

if N < 20:
    large_problem = False
else:
    large_problem = True


for generation in range(GENERATIONS):

    if large_problem:
        parents = rank_selection(population)
    else:
        parents = tournament_selection(population)

    population = create_new_population(
        parents,
        large_problem
    )

    population.sort(key=route_distance)

    if generation % 20 == 0:

        print(
            "Generation:",
            generation,
            "Best Distance:",
            round(route_distance(population[0]), 2)
        )


best_route = min(
    population,
    key=route_distance
)


print("\nBest Route:")

print(best_route + [best_route[0]])


print("\nMinimum Distance:")

print(round(route_distance(best_route), 2))
```
